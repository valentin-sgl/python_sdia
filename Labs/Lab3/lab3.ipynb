## <a name="ex1">Exercise 1: Brownian motion</a>

This first exercise consists in generating a Brownian motion on the closed unit ball $\mathcal{B}(\mathbf{0}, 1) = \{ \mathbf{x} \mid \Vert \mathbf{x} \Vert  \leq 1\}$, focusing first on the 2-D case. The Brownian motion is a random walk with independent, identically distributed Gaussian increments, appearing for instance in thermodynamics and statistical mechanics (to model the evolution of a large particle in a medium composed of a large number of small particles, ...). It is also connected to the diffusion process (Einstein).

Let $N \in \mathbb{N}^*$, $\delta > 0$, and $\mathbf{x} = (x_1, x_2) \in  \mathcal{B}(\mathbf{0}, 1)$. The first $N$ steps of a 2-D discrete-time Brownian motion $W$ can be generated as follows

\begin{align*}
    W_0 &= \mathbf{x}, \\
    %
    (\forall n \in \{1, \dotsc, N-1 \}), \quad W_n &= W_{n−1} + \sqrt{\delta} G_n, \quad G_n \sim \mathcal{N}(\mathbf{0}, \mathbf{I}),
\end{align*}

where $\mathcal{N}(\mathbf{0}, \mathbf{I})$ is a Gaussian distribution with mean $\mathbf{0}$ and identity covariance matrix.

2. Implement a function `brownian_motion(niter, x, step, rng)` which

    - simulates $W$ until it reaches the boundary of $\mathcal{B}(\mathbf{0}, 1)$, using a maximum of $N$ iterations (`niter`), a starting point $\mathbf{x} \in \mathcal{B}(\mathbf{0}, 1)$ (`x`) and step-size $\delta$ (`step`);
    - interpolates linearly between the two last positions to determine the points $W^*$ where the trajectory crosses the boundary (if applicable);
    - returns both the whole random walk $W$ and, if appropriate, the point at the intersection between the last segment of the trajectory and $\mathcal{B}(\mathbf{0}, 1)$.
 
> Hint: 
> - you can easily derive a closed form expression for $W^*$, observing that $\Vert W^* \Vert^2= 1$ and $W^* \in [W_{n-1}, W_n]$. 
> - you can also take a look at [`np.roots`](https://numpy.org/doc/stable/reference/generated/numpy.roots.html?highlight=roots#numpy.roots) if needed.

> Recall of the Linear Interpolation (LERP) for $n$-dimensional vectors:
> - Clearly, $\vec{D}=\alpha \vec{C}$ with $\alpha \in [0, 1]$ and hence $\vec{P}-\vec{B}= \alpha (\vec{A}-\vec{B})$ which is equivalent to $\vec{P}= (1-\alpha) \vec{B} + \alpha \vec{A}$. 
![alternatvie text](img/for_Course.png)

**Answer:**

````python
import numpy as np


def brownian_motion(niter, x, step, rng):
    """
    Parameters
    ----------
    niter : int
        Maximum number of Brownian steps.
    x : array_like
        Initial position in the unit disk.
    step : float
        Variance parameter of the Gaussian increment (delta).
    rng : numpy.random.Generator
        Random number generator.

    Returns
    -------
    W : np.ndarray
        Array of positions along the trajectory, including the starting point.
    W_star : np.ndarray or None
        Intersection point with the boundary if it is reached before niter,
        otherwise None.
    """
    x = np.asarray(x, dtype=float)
    if x.shape != (2,):
        x = x.reshape(2)

    if np.linalg.norm(x) >= 1.0:
        return np.array([x], dtype=float), x.copy()

    W = [x.copy()]
    current = x.copy()

    for _ in range(niter):
        increment = np.sqrt(step) * rng.standard_normal(2)
        next_pos = current + increment
        W.append(next_pos.copy())

        if np.linalg.norm(next_pos) >= 1.0:
            prev = current
            d = next_pos - prev

            # Solve ||prev + t * d||^2 = 1 with t in [0, 1]
            a = d @ d
            b = 2 * prev @ d
            c = prev @ prev - 1.0

            if np.isclose(a, 0.0):
                W_star = prev.copy()
            else:
                roots = np.roots([a, b, c])
                valid = roots[(np.isclose(roots.imag, 0.0))
                              & (roots.real >= 0.0)
                              & (roots.real <= 1.0)]
                if len(valid) == 0:
                    W_star = None
                else:
                    t = valid[0].real
                    W_star = prev + t * d

            return np.array(W, dtype=float), W_star

        current = next_pos

    return np.array(W, dtype=float), None
````

````python
import numpy as np


def brownian_motionv2(niter, x, step, rng):
    """
    Paramètres :
    niter : int
        Nombre maximum d'itérations.
    x : array_like
        Position initial
    step : float
        pas.
    rng : numpy.random.Generator
        Générateur de nombres aléatoires.

    Returns
    W : np.ndarray
        Tableau des différentes positions, dont le point de départ.
    W_star : np.ndarray or None
        Point d'intersection avec la frontière s'il est atteint avant niter,
        sinon None.
    """

    # On vérifie que x est bien dans de dimension 2. Si ce n'est pas le cas on le redimensionne.
    x = np.asarray(x, dtype=float)
    if x.shape != (2,):
        x = x.reshape(2)

    # Si x n'appartient pas à B(0,1), on ne lance pas la procédure;
    if np.linalg.norm(x) >= 1.0:
        return np.array([x], dtype=float), x.copy()

    # On initialise la liste des vecteurs à renvoyer, qui contient déjà le point de départ
    W = [x.copy()]
    W_n = x.copy()

    # On a un nombre d'itérations limité
    for _ in range(niter):
        # On calcule W_n+1
        G = rng.standard_normal(2) # Calcul de la gaussienne de dimension 2
        W_n = W_n + G*np.sqrt(step)
        W.append(W_n.copy())
        
        # On vérifie si W_n appartient encore à B(0,1).
        # Si W_n n'appartient plus à B(0,1), on va chercher W* le vecteur à la frontière.
        if np.linalg.norm(W_n) >= 1 :
            W_in = W[-2]
            W_out = W_n

            # Rappel : il existe t appartient [0,1] tel que :
            # W* = t*W_out + (1-t)*W_in

            # On a ||W*||^2 = 1.
            # En remplaçant W* dans l'expression précédente, on obtient :
            # ||W_out-W_in||^2 * t^2 - 2 <W_in,W_out-W_in> * t + ||W_out||^2 - 1 = 0
            # On résout ce système polynomiale pour trouver t (et en déduire W*).
            d = W_out-W_in
           
            a = d@d
            b = 2*W_in@d
            c = W_in@W_in - 1

            if np.isclose(a, 0.0) : # cas où W_in est sur la frontière de B(0,1) (||W_in|| = 1)
                W_lim = W_in.copy()
            else :
                roots = np.roots([a, b, c]) # solutions de l'équation du second degré
                # On conserve uniquement la solution comprise entre 0 et 1.
                t_valide = roots[(np.isclose(roots.imag, 0.0))
                              & (roots.real >= 0.0)
                              & (roots.real <= 1.0)]
                if len(t_valide) == 0 : # pas de solution comprise entre 0 et 1
                    W_lim = None
                else:
                    t = t_valide[0].real
                    W_lim = W_out*t + (1-t)*W_in

            return np.array(W, dtype=float), W_lim # On arrête d'itérer


    return np.array(W, dtype=float), None
````

````python
# Exemple d'utilisation
rng = np.random.default_rng(0)
W, W_lim = brownian_motion(1000, np.array([0.2, 0.4]), 1e-2, rng)
print(W[:5])
print(W_star)
````

5. [Bonus] Generalize the procedure to a $M$-dimensional Brownian motion, $M > 2$.

**Answer:**

````python
def brownian_motion_mdim(niter, x, step, rng):
    """
    Parameters
    ----------
    niter : int
        Maximum number of steps.
    x : array_like
        Starting point in R^M with ||x|| <= 1.
    step : float
        Step-size parameter delta.
    rng : numpy.random.Generator
        Random number generator.

    Returns
    -------
    W : np.ndarray
        Array of shape (k, M), all positions visited before exit.
    W_star : np.ndarray or None
        Intersection point with the ball boundary if it exits before niter.
    """
    x = np.asarray(x, dtype=float)
    M = x.size

    if np.linalg.norm(x) >= 1.0:
        return x.reshape(1, M), x.copy()

    W = [x.copy()]
    current = x.copy()

    for _ in range(niter):
        increment = np.sqrt(step) * rng.standard_normal(M)
        next_pos = current + increment
        W.append(next_pos.copy())

        if np.linalg.norm(next_pos) >= 1.0:
            prev = current
            d = next_pos - prev

            # Solve ||prev + t d||^2 = 1  with  t in [0, 1]
            a = d @ d
            b = 2 * prev @ d
            c = prev @ prev - 1.0

            if np.isclose(a, 0.0):
                W_star = prev.copy()
            else:
                discriminant = b**2 - 4*a*c
                if discriminant < 0:
                    W_star = None
                else:
                    roots = np.roots([a, b, c])
                    valid = roots[(np.isclose(roots.imag, 0.0))
                                  & (roots.real >= 0.0)
                                  & (roots.real <= 1.0)]
                    if len(valid) == 0:
                        W_star = None
                    else:
                        t = valid[0].real
                        W_star = prev + t * d

            return np.array(W, dtype=float), W_star

        current = next_pos

    return np.array(W, dtype=float), None
````

````python
#Example for M = 3
rng = np.random.default_rng(0)
x0 = np.array([0.2, -0.3, 0.1])
W, W_star = brownian_motion_mdim(1000, x0, 1e-2, rng)
print(W[:5])
print(W_star)
````

2\. Let $\mathcal{F}$ denote the 2D discrete Fourier transform. Compute $|\mathcal{F}(\mathbf{X})|^2$, the spectrum of the image $\mathbf{X} \in \mathbb{R}^{M_1 \times N_1}$ (i.e., the term-wise squared absolute value of its Fourier transform) loaded in 1. Display the result in logarithmic scale.

a) In this representation, where is the pixel of the spectrum associated with the null frequency located?
    
b) Take a look at the documentation of `np.fft.fftshift`. Use it to ensure that the null frequency is located at the center of the image. 

**Answer:**

````python
# On calcule le spectre de l'image X
F_X = np.fft.fft2(X)

Spectrum_X = np.abs(F_X)**2
Spectrum_X_log = np.log(1+Spectrum_X) #on ajoute 1 pour éviter le cas limite log(0)

# On calcule le spectre de l'image X décalée afin d'avoir la fréquence nulle centrée
F_X_centre = np.fft.fftshift(F_X)
Spectrum_X_centre_log = np.log(1+np.abs(F_X_centre)**2)


# On recherche le premier pixel dont la fréquence est nulle dans le spectre de X.
frequence_nulle = np.max(Spectrum_X)
M1, N1 = Spectrum_X.shape

i = 0
trouve = False
while i < M1 and not trouve:
    j = 0
    while j < N1 and not trouve:
        if Spectrum_X[i, j] == frequence_nulle:
            position_frequence_nulle = (i, j)
            trouve = True
        j += 1
    i += 1

print(f"Pixel de la fréquence nulle trouvé aux coordonnées : {position_frequence_nulle}")

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(Spectrum_X_log, cmap='gray')
plt.title("Spectre non centré")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(Spectrum_X_centre_log, cmap='gray')
plt.title("Spectre centré (avec fftshift)")
plt.axis('off')
plt.show()
````

3\. 
    a) Create a function `ideal_lowpass_filter` to filter $\mathbf{X}$ by an ideal low-pass filter. The filter preserves Fourier coefficients associated to frequencies below a cutoff specified in each direction ($\mathbf{f}_c = (f_{c,y}, f_{c,x})$), and sets others to zero. For simplicity, $f_{c,y}$ and $f_{c,x}$ can be expressed as a number of samples to be kept along each dimension (e.g., $\mathbf{f}_c = (50,50)$).

b) Display the filtered image for 2 different values of $\mathbf{f}_c$. What do you observe as the cutoff frequencies increase?
    
> Warning: beware the type of the array after `np.fft.fft2`, do not hesitate to specify the type if you make copies from this array
> ```python
> a = np.zeros((2,2), dtype=np.complex)
> ...
> ```

**Answer:**

````python
def ideal_lowpass_filter(image, f_c) :

    f_cx, f_cy = f_c

    # On importe l'image, que l'on exprime dans une matrice 2D, et plus en RGB (matrice 3D)
    img = plt.imread(image)
    img = np.mean(img, axis = 2)

    # On calcule la transformée de Fourier avec la fréquence nulle centrée dans l'image
    F_img = np.fft.fftshift(np.fft.fft2(img))

    #On crée le filtre rempli de 0
    filtre = np.zeros(F_img.shape, dtype=np.complex128) # np.complex128 et non np.complex (plus dans les versions récentes de numpy)

    M1, N1 = F_img.shape

    # Grâce à np.fft.fftshift, nos plus basses fréquences sont centrées dans l'image
    # Comme f_c correspond à la taille de l'échantillon (et non une fréquence pure), on définit les limites de la zone à conserver autour du centre
    centre_x = N1//2
    centre_y = M1//2
    x_min = centre_x - f_cx//2
    x_max = centre_x + f_cx//2
    y_min = centre_y - f_cy//2
    y_max = centre_y + f_cy//2

    # On ajoute 1 dans la matrice "filtre" pour les fréquences qui ont passé le filtre passe-bas_
    filtre[y_min:y_max, x_min:x_max] = 1

    # Produit d'Hadamard pour conserver uniquement les fréquences qui ont passé le filtre
    F_img_filtree = F_img*filtre

    # Transformée inverse de Fourier
    img_filtree = np.real(np.fft.ifft2(np.fft.ifftshift(F_img_filtree)))

    return img_filtree


f_c1 = [50,50]
f_c2 = [25,25]

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(X, cmap='gray')
plt.title("Image non filtrée")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(ideal_lowpass_filter('img/boat.png', f_c1), cmap='gray')
plt.title(f"Image filtrée (avec fc = {f_c1})")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(ideal_lowpass_filter('img/boat.png', f_c2), cmap='gray')
plt.title(f"Image filtrée (avec fc = {f_c2})")
plt.axis('off')

plt.show()
````

Lorsque l'on diminue $\mathbf{f}_c$ ; l'image devient de plus en plus floue ; les détails disparaissent.

4\. Let $\mathbf{H} \in \mathbb{R}^{M_2\times N_2}$ be a 2-D Gaussian kernel, obtained as the outer product of two 1-D Gaussian windows $\mathbf{w}_y \in \mathbb{R}^{M_2}$ and $\mathbf{w}_x \in \mathbb{R}^{N_2}$, of standard deviation $\sigma_y = 10$ and $\sigma_x = 10$, respectively:

\begin{equation}
    \mathbf{H} = \mathbf{w}_y \mathbf{w}_x^T.
\end{equation}

Let $M = M_1+M_2-1$ and $N =  N_1+N_2-1$. From the discrete convolution theorem, the linear convolution between $\mathbf{H}$ and $\mathbf{X}$ can be computed as follows

\begin{equation}
    \mathbf{X} \star \mathbf{H} = \mathcal{F}^{-1} \Big( \mathcal{F}\big(P_1(\mathbf{X})\big) \odot \mathcal{F}\big(P_2(\mathbf{H})\big) \Big) \in \mathbb{R}^{M\times N},
\end{equation}

where $P_i: \mathbb{R}^{M_i \times N_i} \rightarrow \mathbb{R}^{M \times N}$, $i \in \{1, 2\}$, are 0-padding operators, $\odot$ is the Hadamard (= term-wise) product, $\mathcal{F}^{-1}$ is the 2D discrete inverse Fourier transform.

Compute and display $\mathbf{X} \star \mathbf{H}$, for $M_2 = N_2 = 10$. What do you observe?

> Hint: 
> - the usual 0-padding procedure in image space consists in appending trailing zeros. For instance (in 1D), 0-padding a vector $\mathbf{x} \in \mathbb{R}^N_1$ to the size $N>N_1$ corresponds to creating the vector
\begin{bmatrix}
\mathbf{x} \\
\mathbf{0}_{N-N_1}
\end{bmatrix}
> - since the input images are real, $\mathcal{F}(\mathbf{x})$ and $\mathcal{F}(\mathbf{h})$ are Hermitian symmetric. In this case, a more efficient version of `np.fft.fft2` can be used, computing only quarter of the Fourier coefficients (half of the Fourier coefficients in each direction): [`np.fft.rfft2`](https://numpy.org/doc/stable/reference/generated/numpy.fft.rfft2.html?highlight=rfft#numpy.fft.rfft2). Its inverse, [`np.fft.irfft2`](https://numpy.org/doc/stable/reference/generated/numpy.fft.irfft2.html#numpy.fft.irfft2), also ensures that the output is real;
> - the 2D Gaussian window can be generated as the outer product of two 1D Gaussian windows (one window for each dimension);
> - you can take a look at [scipy.signal.windows.gaussian](https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.windows.gaussian.html#scipy.signal.windows.gaussian) and [np.newaxis](https://numpy.org/doc/stable/reference/constants.html?highlight=newaxis#numpy.newaxis) (or `np.reshape` or built-in `np.outer`).

**Answer:**

````python
from scipy.signal.windows import gaussian

M2, N2 = 10, 10
M, N = M1+M2-1, N1+N2-1

sigma_y, sigma_x = 10, 10

w_y = gaussian(M2, sigma_y)
w_x = gaussian(N2, sigma_x)
w_y_colonne = w_y[:, np.newaxis] # transforme w_y en une matrice de dimension (10, 1)
w_x_ligne = w_x[np.newaxis, :] # transforme w_x en une matrice de dimension (1, 10)

# Le produit d'une matrice (10, 1) avec une matrice (1, 10) est bien une matrice de dimension (10, 10)
H = w_y_colonne*w_x_ligne

# On crée les 0-paddings operators
P_1_X = np.zeros([M,N])
P_2_H = np.zeros([M,N])
P_1_X[:M1, :N1] = X
P_2_H[:M2, :N2] = H

# On calcule les transformées de Fourier
F_P_X = np.fft.rfft2(P_1_X)
F_P_H = np.fft.rfft2(P_2_H)

# Rappel du produit d'Hadamard : soit A et B deux matrices de même dimension, si C = A*B alors C[i][j] = A[i][j]*B[i][j] 
Hadamard_product = F_P_X*F_P_H

# On calcule la transformée inverse de Fourier
conv_X_H = np.fft.irfft2(Hadamard_product)


plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(X, cmap='gray')
plt.title("Image d'origine")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(conv_X_H, cmap='gray')
plt.title("Image après convolution linéaire avec H")
plt.axis('off')
plt.show()
````

On observe que l'image passée dans par le produit de convolution est floue, comme les images traitées par le filtre passe-bas dans la question précédente. On peut faire que le noyau gaussien a agi comme un filtre passe-bas.

On remarque aussi l'apparition d'un cadre noir autour de l'image filtrée. En bas et à droite, ces bordures sont dues aux 0 ajoutés via le 0-padding.
En haut et à drotie, les bordures sont dues au noyau gaussien qui contient de très faibles valeurs à ses extremités (rappel : la gaussienne est symétrique avec unpic d'intensité maximale en son centre).
